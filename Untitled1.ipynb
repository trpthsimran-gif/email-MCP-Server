```python
#Cell 4 — Connect to Gmail (opens browser login)
SCOPES = ['https://www.googleapis.com/auth/gmail.modify']

def connect_gmail():
    creds = None
    
    # Use saved token if exists
    if os.path.exists('token.json'):
        creds = Credentials.from_authorized_user_file('token.json', SCOPES)
    
    # Login if no valid token
    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
            print("✅ Token refreshed automatically!")
        else:
            # This opens your browser to login
            flow = InstalledAppFlow.from_client_secrets_file('credentials.json', SCOPES)
            creds = flow.run_local_server(port=0)
            print("✅ Login successful!")
        
        # Save token so you never login again
        with open('token.json', 'w') as f:
            f.write(creds.to_json())
        print("✅ Token saved to token.json")
    
    else:
        print("✅ Already logged in!")
    
    service = build('gmail', 'v1', credentials=creds)
    print("🎉 Gmail connected successfully!")
    return service

# Run it — browser will open for login first time
gmail = connect_gmail()
```

```python
#Cell 6 — Send an email
def send_email(to, subject, body):
    print(f"📤 Sending email to {to}...")
    
    message = MIMEText(body)
    message['to'] = to
    message['subject'] = subject
    
    # Gmail needs base64 encoded message
    raw = base64.urlsafe_b64encode(message.as_bytes()).decode()
    
    result = gmail.users().messages().send(
        userId='me',
        body={'raw': raw}
    ).execute()
    
    print(f"✅ Email sent successfully!")
    print(f"   Message ID: {result['id']}")
    return result

# Test — send yourself an email
send_email(
    to="your@gmail.com",          # ← change to your email
    subject="Hello from Jupyter!",
    body="This email was sent from my MCP server built in Jupyter! 🎉"
)
```

```python
#Cell 9 — Export to index.py (the MCP server)
code = '''import asyncio
import json
from gmail import read_emails, send_email, search_emails

# Simple MCP server using stdio
import sys

def handle_tool_call(tool_name, params):
    if tool_name == "read_emails":
        count = params.get("count", 5)
        emails = read_emails(count)
        return json.dumps(emails, indent=2)
    
    elif tool_name == "send_email":
        result = send_email(params["to"], params["subject"], params["body"])
        return result
    
    elif tool_name == "search_emails":
        emails = search_emails(params["query"])
        return json.dumps(emails, indent=2)
    
    else:
        return f"Unknown tool: {tool_name}"

def main():
    while True:
        line = sys.stdin.readline()
        if not line:
            break
        try:
            request = json.loads(line)
            tool = request.get("tool")
            params = request.get("params", {})
            result = handle_tool_call(tool, params)
            response = {"result": result}
        except Exception as e:
            response = {"error": str(e)}
        
        print(json.dumps(response), flush=True)

if __name__ == "__main__":
    main()
'''

with open('index.py', 'w') as f:
    f.write(code)

print("✅ index.py created!")
print()
print("🎉 All files ready!")
print()
print("📁 Your folder now contains:")
for f in os.listdir('.'):
    print("  ", f)
```

```python
import time

def get_email_body(payload):
    """Extract actual text content from email"""
    body = ""
    if 'parts' in payload:
        for part in payload['parts']:
            if part['mimeType'] == 'text/plain':
                data = part['body'].get('data', '')
                if data:
                    body = base64.urlsafe_b64decode(data).decode('utf-8', errors='ignore')
                    break
    else:
        data = payload.get('body', {}).get('data', '')
        if data:
            body = base64.urlsafe_b64decode(data).decode('utf-8', errors='ignore')
    return body[:500]


def read_emails(count=5, retries=3):
    """Read emails with automatic retry on connection error"""
    global gmail          # ← moved to TOP of function

    for attempt in range(retries):
        try:
            print(f"📬 Fetching last {count} emails...\n")

            results = gmail.users().messages().list(
                userId='me',
                maxResults=count
            ).execute()

            messages = results.get('messages', [])

            if not messages:
                print("No emails found.")
                return []

            emails = []
            for i, msg in enumerate(messages, 1):
                detail = gmail.users().messages().get(
                    userId='me',
                    id=msg['id']
                ).execute()

                headers = detail['payload']['headers']

                subject = next((h['value'] for h in headers if h['name'] == 'Subject'), '(no subject)')
                sender  = next((h['value'] for h in headers if h['name'] == 'From'), 'Unknown')
                date    = next((h['value'] for h in headers if h['name'] == 'Date'), '')
                body    = get_email_body(detail['payload'])

                emails.append({
                    'id':      msg['id'],
                    'subject': subject,
                    'from':    sender,
                    'date':    date,
                    'body':    body
                })

                print(f"{i}. 📧 {subject}")
                print(f"   From : {sender}")
                print(f"   Date : {date}")
                print(f"   Body : {body[:100]}...")
                print()

            return emails

        except Exception as e:
            if 'SSLEOFError' in str(type(e).__name__) or 'SSL' in str(e):
                print(f"⚠️ Connection dropped (attempt {attempt+1}/{retries})")
                print(f"   Waiting 3 seconds and retrying...")
                time.sleep(3)
                try:
                    gmail = connect_gmail()
                    print("✅ Reconnected!")
                except:
                    pass
            else:
                print(f"❌ Error: {e}")
                raise e

    print("❌ Failed after 3 attempts. Check your internet connection.")
    return []


# Test
my_emails = read_emails(3)
```

```python
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText

def reply_in_same_thread(email_id, reply_body):
    """
    Reply inside same email thread
    No new email created
    """
    global gmail
    
    print(f"🔍 Finding original email...")
    
    # Step 1 - Get original email
    original = gmail.users().messages().get(
        userId='me',
        id=email_id,
        format='full'
    ).execute()
    
    headers   = original['payload']['headers']
    thread_id = original['threadId']
    
    # Step 2 - Extract headers
    original_from    = ''
    original_subject = ''
    message_id       = ''
    references       = ''
    
    for h in headers:
        name = h['name'].lower()
        if name == 'from':
            original_from = h['value']
        elif name == 'subject':
            original_subject = h['value']
        elif name == 'message-id':
            message_id = h['value']
        elif name == 'references':
            references = h['value']
    
    print(f"✅ Found email!")
    print(f"   Replying to  : {original_from}")
    print(f"   Subject      : {original_subject}")
    print(f"   Thread ID    : {thread_id}")
    
    # Step 3 - Build subject
    if original_subject.lower().startswith('re:'):
        reply_subject = original_subject
    else:
        reply_subject = 'Re: ' + original_subject
    
    # Step 4 - Build references chain
    if references:
        new_references = references + ' ' + message_id
    else:
        new_references = message_id
    
    # Step 5 - Build reply message
    msg = MIMEMultipart()
    msg['To']      = original_from
    msg['Subject'] = reply_subject
    
    if message_id:
        msg['In-Reply-To'] = message_id
        msg['References']  = new_references
    
    msg.attach(MIMEText(reply_body, 'plain'))
    
    # Step 6 - Encode and send
    raw = base64.urlsafe_b64encode(
        msg.as_bytes()
    ).decode('utf-8')
    
    result = gmail.users().messages().send(
        userId='me',
        body={
            'raw': raw,
            'threadId': thread_id    # ← keeps in same thread
        }
    ).execute()
    
    # Step 7 - Verify
    if result['threadId'] == thread_id:
        print(f"\n✅ Reply sent in SAME thread!")
    else:
        print(f"\n⚠️ Different thread - check Gmail")
    
    print(f"   Message ID : {result['id']}")
    return result


print("✅ reply_in_same_thread function is ready!")
```

```python
index_code = '''import sys
import json
import os

sys.path.insert(0, r"C:\\Users\\thema\\email-mcp-server")
os.chdir(r"C:\\Users\\thema\\email-mcp-server")

from gmail import read_emails, send_email, search_emails, reply_in_thread

def handle_request(request):
    method = request.get("method", "")
    req_id = request.get("id")

    if method == "initialize":
        return {
            "jsonrpc": "2.0",
            "id": req_id,
            "result": {
                "protocolVersion": "2024-11-05",
                "capabilities": {"tools": {}},
                "serverInfo": {"name": "email", "version": "3.0.0"}
            }
        }

    elif method == "tools/list":
        return {
            "jsonrpc": "2.0",
            "id": req_id,
            "result": {
                "tools": [
                    {
                        "name": "read_emails",
                        "description": "Read recent emails from Gmail with full body content. Always use this first to get email IDs before replying.",
                        "inputSchema": {
                            "type": "object",
                            "properties": {
                                "count": {
                                    "type": "number",
                                    "description": "Number of emails to fetch"
                                }
                            }
                        }
                    },
                    {
                        "name": "reply_in_thread",
                        "description": "Reply to an existing email INSIDE the same conversation thread. No new email is created. ALWAYS use this when user wants to reply to a received email. Never use send_email for replies.",
                        "inputSchema": {
                            "type": "object",
                            "properties": {
                                "email_id": {
                                    "type": "string",
                                    "description": "The ID of the email to reply to. Get this from read_emails first."
                                },
                                "reply_body": {
                                    "type": "string",
                                    "description": "The reply message text"
                                }
                            },
                            "required": ["email_id", "reply_body"]
                        }
                    },
                    {
                        "name": "send_email",
                        "description": "Send a BRAND NEW email. Only use this for new emails, never for replies.",
                        "inputSchema": {
                            "type": "object",
                            "properties": {
                                "to":      {"type": "string", "description": "Recipient email address"},
                                "subject": {"type": "string", "description": "Email subject"},
                                "body":    {"type": "string", "description": "Email body"}
                            },
                            "required": ["to", "subject", "body"]
                        }
                    },
                    {
                        "name": "search_emails",
                        "description": "Search emails using Gmail search syntax",
                        "inputSchema": {
                            "type": "object",
                            "properties": {
                                "query": {
                                    "type": "string",
                                    "description": "Gmail search e.g. from:john, subject:invoice, is:unread"
                                }
                            },
                            "required": ["query"]
                        }
                    }
                ]
            }
        }

    elif method == "tools/call":
        tool_name = request.get("params", {}).get("name", "")
        args      = request.get("params", {}).get("arguments", {})

        try:
            if tool_name == "read_emails":
                count  = int(args.get("count", 5))
                result = read_emails(count)
                text   = json.dumps(result, indent=2)

            elif tool_name == "reply_in_thread":
                text = reply_in_thread(
                    args["email_id"],
                    args["reply_body"]
                )

            elif tool_name == "send_email":
                text = send_email(
                    args["to"],
                    args["subject"],
                    args["body"]
                )

            elif tool_name == "search_emails":
                result = search_emails(args["query"])
                text   = json.dumps(result, indent=2)

            else:
                text = f"Unknown tool: {tool_name}"

            return {
                "jsonrpc": "2.0",
                "id": req_id,
                "result": {
                    "content": [{"type": "text", "text": text}]
                }
            }

        except Exception as e:
            return {
                "jsonrpc": "2.0",
                "id": req_id,
                "result": {
                    "content": [{"type": "text", "text": f"Error: {str(e)}"}]
                }
            }

    elif method == "notifications/initialized":
        return None

    else:
        return {
            "jsonrpc": "2.0",
            "id": req_id,
            "error": {"code": -32601, "message": f"Method not found: {method}"}
        }


def main():
    while True:
        try:
            line = sys.stdin.readline()
            if not line:
                break
            line = line.strip()
            if not line:
                continue
            request  = json.loads(line)
            response = handle_request(request)
            if response is not None:
                print(json.dumps(response), flush=True)
        except json.JSONDecodeError:
            continue
        except Exception as e:
            sys.stderr.write(f"Error: {e}\\n")
            sys.stderr.flush()

if __name__ == "__main__":
    main()
'''

with open('index.py', 'w') as f:
    f.write(index_code)

print("✅ index.py updated!")
print()
print("📁 Files in folder:")
for f in os.listdir('.'):
    print("  ", f)
```
